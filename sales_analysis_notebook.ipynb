# Automatic Sales Analytics — Exploration Notebook

This notebook lets you explore a sales file manually (step by step) using
the same `data_cleaner.py` and `analyzer.py` modules that power the
Streamlit dashboard.

**How to use:**
1. Put your sales CSV/Excel file in this same folder.
2. Change `FILE_NAME` below to match your file.
3. Run each cell in order (Shift + Enter).


```
# Step 1: Import the project's modules and libraries
import pandas as pd
import plotly.express as px

from data_cleaner import clean_pipeline
from analyzer import (
    summary_kpis, revenue_over_time, top_items,
    growth_rate, category_breakdown, generate_text_report
)

pd.set_option("display.max_columns", None)
```

```
# Step 2: Load and auto-clean your file
FILE_NAME = "sample_sales_data.csv"  # <-- change this to your file name

df, column_map = clean_pipeline(FILE_NAME)

print("Detected columns:", column_map)
print("Rows after cleaning:", len(df))
df.head()
```

## Step 3: Key metrics
These are the headline numbers: total revenue, order count, average
order value, and total units sold.

```
kpis = summary_kpis(df, column_map)
kpis
```

## Step 4: Revenue trend over time

```
ts = revenue_over_time(df, column_map, freq="ME")  # 'D'=daily, 'W'=weekly, 'ME'=monthly
fig = px.line(ts, x="period", y="revenue", markers=True, title="Monthly Revenue")
fig.show()

print("Growth from first to last period:", growth_rate(ts), "%")
```

## Step 5: Top products, regions, customers

```
top_products = top_items(df, column_map, role="product", n=10)
top_products
```

```
fig = px.bar(top_products, x="Revenue", y=top_products.columns[0], orientation="h",
             title="Top 10 Products by Revenue")
fig.update_layout(yaxis={"categoryorder": "total ascending"})
fig.show()
```

## Step 6: Category breakdown

```
cat = category_breakdown(df, column_map)
if not cat.empty:
    fig = px.pie(cat, names="Category", values="Revenue", title="Revenue by Category")
    fig.show()
else:
    print("No category column detected in this file.")
```

## Step 7: Auto-generated written report

```
report = generate_text_report(df, column_map)
print(report)

# Save it to a file
with open("reports/sales_report.txt", "w") as f:
    f.write(report)
print("\nSaved to reports/sales_report.txt")
```
